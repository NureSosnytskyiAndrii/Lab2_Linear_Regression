```python
import pandas as pd
from sklearn.model_selection import train_test_split

sales = pd.read_csv('kc_house_data.csv')

train_data, test_data = train_test_split(sales, test_size=0.8, random_state=0)

print(f"Навчальна вибірка: {len(train_data)} рядків")
print(f"Тестова вибірка:    {len(test_data)} рядків")
```

```python
import numpy as np

def simple_linear_regression(input_feature, output):
    x = np.array(input_feature)
    y = np.array(output)
    n = len(x)
    
    x_mean = x.mean()
    y_mean = y.mean()
    
    numerator = (x * y).sum() - n * x_mean * y_mean
    
    denominator = (x * x).sum() - n * x_mean**2
    
    slope = numerator / denominator
    intercept = y_mean - slope * x_mean
    
    return intercept, slope

test_feature = np.arange(5)
test_output = 1 + 1 * test_feature

test_intercept, test_slope = simple_linear_regression(test_feature, test_output)

print(f"Перевірка:")
print(f"Intercept: {test_intercept:.1f} (очікується 1.0)")
print(f"Slope:     {test_slope:.1f} (очікується 1.0)")
```

```python
def get_regression_predictions(input_feature, intercept, slope):
    predictions = intercept + slope * np.array(input_feature)
    return predictions

def get_residual_sum_of_squares(input_feature, output, intercept, slope):
    predictions = get_regression_predictions(input_feature, intercept, slope)
    residuals = np.array(output) - predictions
    RSS = (residuals**2).sum()
    return RSS

rss_check = get_residual_sum_of_squares(np.arange(5), 1 + 1*np.arange(5), 1.0, 1.0)
print(f"Перевірка RSS на тестових даних: {rss_check:.1f} (очікується 0.0)")

sqft_intercept, sqft_slope = simple_linear_regression(train_data['sqft_living'], train_data['price'])

rss_train = get_residual_sum_of_squares(train_data['sqft_living'], train_data['price'], sqft_intercept, sqft_slope)
rss_test = get_residual_sum_of_squares(test_data['sqft_living'], test_data['price'], sqft_intercept, sqft_slope)

house_2650_pred = get_regression_predictions(2650, sqft_intercept, sqft_slope)
```
